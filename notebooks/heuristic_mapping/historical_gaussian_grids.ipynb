# Historical Gaussian Grid Reference

Reproduce the two 120-episode broad grids, the separate 120-episode distinct equal-outcome grid, and the three 1,200-episode focused confirmations using their identified historical scientific source. The historical solver forced equal allocation for exchangeable beliefs; the current solver no longer does. This notebook does not change the current model.

The runner extracts only scientific source from Git commit `cc73e14f25a5c9eff2da12838467de54454464ca` into the ignored output directory. Use a full Git clone; for a shallow clone, fetch this commit first. No private result files are required. Numerical libraries can affect exact floating-point results; this is a source-and-recipe reference, not a promise of bitwise equality across runtimes.


```python
from pathlib import Path
import subprocess
import sys

ROOT = next(p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents) if (p / 'scripts/reproduce_gaussian_reference.py').is_file())
RUN = False
MODE = 'full'  # smoke: one environment/episode, not scientific evidence
WORKERS = 1  # Increase to the useful CPU capacity available to this process.
RECIPES = ['near_50_50', 'equal_outcome', 'equal_outcome_distinct_serious']
# Optional confirmations: near_50_50_focused, equal_outcome_focused, equal_outcome_distinct_focused
OUTPUT = ROOT / 'results/heuristic_mapping/historical_gaussian' / MODE
assert MODE in ('smoke', 'full')
```

## Scientific settings

All full recipes use the historical serious preset, 500 VOI draws and common observation streams with at least 200 observations per person. Base environment seed: 11; allocation grid: 31; utility draws: 100. Broad near-50/50 and equal-outcome grids use 64 chunks each; distinct serious uses 32. Focused confirmations use 36, 48 and 64 chunks respectively. Chunking is retained because it can affect historical randomization.

The output directory must be new. Opening this notebook does not start any calculation.


```python
for recipe in RECIPES:
    cmd = [sys.executable, str(ROOT / 'scripts/reproduce_gaussian_reference.py'), '--recipe', recipe, '--mode', MODE, '--workers', str(WORKERS), '--output', str(OUTPUT / recipe)]
    print(' '.join(cmd))
    if RUN:
        subprocess.run(cmd, cwd=ROOT, check=True)
```

## Inspect combined results

Each recipe writes its source identity and status to `reference_recipe.json`. Combined comparison, behavior-profile and candidate CSVs, plus figures, are under `tables/`. Historical output field names are retained. Empty candidate tables are valid negative findings, not missing results.


```python
import pandas as pd
for recipe in RECIPES:
    for path in sorted((OUTPUT / recipe / 'tables').glob('*.csv')):
        print(recipe, path.name)
        if path.stat().st_size:
            display(pd.read_csv(path).head())
```
