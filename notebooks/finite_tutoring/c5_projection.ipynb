# C5 projection: reproducibility draft
Install dependencies using the first cell in a writable checkout and environment before staging a separate read-only source snapshot; then start the kernel in this notebook directory of the staged checkout. For the separate supplied-prefix or saved-result redraw routes, download the inputs from https://drive.google.com/drive/folders/1DueektDCojHq0A2SwlM_XKf-OzOvJqHh and keep them outside Git. Keep inputs and checkpoints outside Git; checkout-local checkpoints must be ignored.

```python
from pathlib import Path
import json
import shlex
import subprocess
import sys

REPOSITORY = Path('../..').resolve()  # Run from the staged checkout.
sys.path.insert(0, str(REPOSITORY))
from src.design_validation.c5_public_authority import ACCEPTED_PREFIX_SHA256
WORK_DIR = Path.home() / 'c5-projection-work'  # Writable, outside the snapshot.
SPEC = 'path/to/single-root-spec.json'
SPEC_SHA256 = ''
LOOKUP = str(REPOSITORY / 'configs' / 'c5_fixed_initial_policy.json')
LOOKUP_SHA256 = '68b2fd36ef99aefcadb78e1831a9167134b62731787e4588d149668699c6d1a0'
PREFIX = 'path/to/staged-supplied-prefix.json'
INPUT_MODE = 'accepted-prefix'  # Or caller-prefix.
PREFIX_SHA256 = ACCEPTED_PREFIX_SHA256 if INPUT_MODE == 'accepted-prefix' else ''  # Override if needed.
OUTPUT = str(WORK_DIR / 'checkpoints')
REPLICATES = '2:2'  # Inclusive; replicate 1 is supplied, not generated.
RUN_PREPARE = False
RUN_SAMPLE = False
RUN_VERIFY = False

runner = [sys.executable, 'scripts/reproduce_c5_projection.py']
prepare = ['--prefix', PREFIX, '--prefix-sha256', PREFIX_SHA256,
           '--lookup', LOOKUP, '--lookup-sha256', LOOKUP_SHA256]
common = ['--mode', 'new-unverified', '--spec', SPEC, '--spec-sha256', SPEC_SHA256,
          '--prefix', PREFIX, '--prefix-sha256', PREFIX_SHA256,
          '--output', OUTPUT, '--replicates', REPLICATES]
for stage, enabled, arguments in [('prepare', RUN_PREPARE, prepare),
                                  ('sample', RUN_SAMPLE, common),
                                  ('verify', RUN_VERIFY, common)]:
    command = [*runner, stage, *arguments]
    print(shlex.join(command))
    if enabled:
        subprocess.run(command, cwd=REPOSITORY, check=True)
```

## Fresh-run draft
This prospective route creates its own first 90 members, then checks that complete prefix before later repetitions. It is under validation: a single test-domain member and synthetic roster checks are not a complete 90-member scientific validation. It does not reconstruct the historical first repetition or authorize statistical inference. Use an isolated read-only checkout and input bundle, with writable output outside them. Keep runtime and source identities unchanged across stages; read-only permissions alone do not prevent concurrent modification.
After the input preparation above, inspect the current byte pins, review them, and explicitly prepare the 90-root roster. The sample, verify and descriptive audit steps are expensive and disabled by default. A failed or interrupted write leaves an operation marker; isolate and inspect it before any manual recovery. Verification creates a temporary lock marker, so the output directory must be writable.

```python
import hashlib
FRESH_STAGED_BUNDLE = Path('path/to/read-only-staged-bundle')
FRESH_MANIFEST = str(FRESH_STAGED_BUNDLE / 'population_manifest.json')
FRESH_MANIFEST_SHA256 = ''  # From preparation_summary.json.
FRESH_PINS = WORK_DIR / 'fresh-pins.json'
FRESH_PREPARED = WORK_DIR / 'fresh-prepared'  # Must be a new directory.
FRESH_CHECKPOINTS = WORK_DIR / 'fresh-checkpoints'
FRESH_RUN_ID = 'choose-a-unique-run-id'
fresh_runner = [sys.executable, 'scripts/fresh_c5_run.py']
RUN_FRESH_PINS = False
pins_command = [*fresh_runner, 'inspect-pins', '--manifest', FRESH_MANIFEST, '--manifest-sha256', FRESH_MANIFEST_SHA256]
print(shlex.join(pins_command))
if RUN_FRESH_PINS:
    completed = subprocess.run(pins_command, cwd=REPOSITORY, check=True, capture_output=True, text=True)
    FRESH_PINS.parent.mkdir(parents=True, exist_ok=True)
    with FRESH_PINS.open('x', encoding='utf-8') as handle:
        handle.write(completed.stdout)
    print(completed.stdout)
    print('Pins SHA256:', hashlib.sha256(FRESH_PINS.read_bytes()).hexdigest())
```

Review the listed byte identities and their sources before enabling preparation. Place the reviewed pins in an isolated read-only snapshot and record their byte SHA256 below. Hashes alone do not authenticate the inputs. Preparation prints a manifest SHA256; retain it with the isolated prepared inputs.

```python
FRESH_REVIEWED_PINS = Path('path/to/read-only-reviewed-pins.json')
FRESH_REVIEWED_PINS_SHA256 = ''  # SHA256 recorded when the pins were reviewed.
RUN_FRESH_PREPARE = False
prepare_command = [*fresh_runner, 'prepare', '--manifest', FRESH_MANIFEST, '--manifest-sha256', FRESH_MANIFEST_SHA256, '--pins', str(FRESH_REVIEWED_PINS), '--pins-sha256', FRESH_REVIEWED_PINS_SHA256, '--output', str(FRESH_PREPARED)]
print(shlex.join(prepare_command))
if RUN_FRESH_PREPARE:
    completed = subprocess.run(prepare_command, cwd=REPOSITORY, check=True, capture_output=True, text=True)
    FRESH_PREPARED_SHA256 = json.loads(completed.stdout)['manifestSHA256']
    print(completed.stdout)
```

Keep the prepared input files isolated and read-only before running the next cell. After restarting the kernel, enter the retained prepared-manifest SHA256 below. Each operation uses the same source and runtime identity; a different runtime is not silently treated as equivalent.

```python
FRESH_PREPARED_SHA256 = globals().get('FRESH_PREPARED_SHA256', '')  # After restart, enter the retained SHA256.
FRESH_REPLICATES = '1:1'  # The first run must include repetition 1.
RUN_FRESH_SAMPLE = RUN_FRESH_VERIFY = RUN_FRESH_AUDIT = False
fresh_args = ['--manifest', str(FRESH_PREPARED / 'input-manifest.json'), '--manifest-sha256', FRESH_PREPARED_SHA256, '--output', str(FRESH_CHECKPOINTS), '--run-id', FRESH_RUN_ID, '--replicates', FRESH_REPLICATES]
for stage, enabled in [('sample', RUN_FRESH_SAMPLE), ('verify', RUN_FRESH_VERIFY), ('audit', RUN_FRESH_AUDIT)]:
    command = [*fresh_runner, stage, *fresh_args]
    print(shlex.join(command))
    if enabled:
        subprocess.run(command, cwd=REPOSITORY, check=True)
```

## Supplied historical prefix
The following separate route uses externally supplied historical data; it does not regenerate that first repetition.
A JSON manifest must contain exactly recipePath, rootsPath, lookupPath, lookupSHA256, latentPath and reportPaths. Paths are relative to the manifest; reportPaths maps each report-width float hex string to its complete accepted product bytes. rootsPath must be the 90 C5 roots produced by the existing public personal-time input generator, not historical outcomes. The adopted integer latent configuration and exact report products must be available separately. The manifest and supplied historical prefix need their own SHA-256 values.
accepted-prefix mode requires the repository's adopted lookup and historical-prefix byte pins; caller-prefix mode checks caller-supplied bytes only. Neither establishes historical dispatch or numerical equivalence.
Use an isolated, immutable staged source checkout and input snapshot throughout each job. File write bits are checked and hashes verified at batch boundaries, but these checks cannot detect temporary modification followed by restoration or prevent a privileged writer replacing files. Keep output paths and Git ignore rules stable during each operation: bounded checks and path rechecks do not eliminate concurrent replacement races. Single-root, population and audit operations share an exclusive output-directory lock. Ordinary audit or verify-only refusals release their locks. Interruptions and failed writing operations retain markers; isolate and review the output rather than deleting a marker blindly.
Sampling and replay are checkpointed per root and future repetition. A full audit requires all 165510 future checkpoints plus the supplied 90-root prefix, replays them independently and uses temporary disk. Set AUDIT_SHARDS to explicit separate checkpoint directories, or leave it empty to audit OUTPUT. The audit acquires a lock marker in each directory and reads, but does not copy or overwrite, checkpoints; directories must be writable for these markers. Missing, duplicate, foreign or incompatible members are rejected. This is expensive and disabled by default; it does not establish historical numerical equivalence or a confidence certificate.

```python
STAGED_BUNDLE = Path('path/to/read-only-staged-bundle')
MANIFEST = str(STAGED_BUNDLE / 'population_manifest.json')
MANIFEST_SHA256 = ''  # manifestSHA256 from preparation_summary.json.
AUDIT_SHARDS = []  # Explicit directories from disjoint, compatible jobs.
RUN_POPULATION_SAMPLE = False
RUN_POPULATION_VERIFY = False
RUN_POPULATION_AUDIT = False
population = ['--mode', 'new-unverified', '--input-mode', INPUT_MODE, '--manifest', MANIFEST,
              '--manifest-sha256', MANIFEST_SHA256, '--prefix', PREFIX,
              '--prefix-sha256', PREFIX_SHA256]
for stage, enabled in [('population-sample', RUN_POPULATION_SAMPLE),
                       ('population-verify', RUN_POPULATION_VERIFY),
                       ('audit', RUN_POPULATION_AUDIT)]:
    if stage == 'audit' and AUDIT_SHARDS:
        arguments = [*population, *(item for path in AUDIT_SHARDS for item in ('--shard-dir', str(path)))]
    elif stage == 'audit':
        arguments = [*population, '--output', OUTPUT]
    else:
        arguments = [*population, '--output', OUTPUT, '--replicates', REPLICATES]
    command = [*runner, stage, *arguments]
    print(shlex.join(command))
    if enabled:
        subprocess.run(command, cwd=REPOSITORY, check=True)
```

## High-price and geometry prerequisites
The public coefficient route below separately produces and independently checks the full 90-root C5 preparation, including the high-price producer and independent checkpoints needed by conditional final. It is not the population input-preparation command, may be expensive, and must use the same checked policy input later if one is supplied. The existing native input generator also makes discarded synthetic ancestry draws; do not describe this route as zero random draws.
For the final CLI, geometry-summary and geometry-certificate must be the raw mathematical objects, not coefficient checkpoint envelopes. The optional source-calling cell loads the checked public preparation and uses certify_supplied_geometry to produce exactly that pair. It performs design-only mathematics, not policy sampling. The standalone reproduce_bounded_prefix.py route independently checks the same planning family but writes a different wrapper; its output is not directly the final CLI certificate. These separate steps, the external prefix, staging, and future-member production mean this is not a one-click historical reproduction.

```python
HIGH_OUTPUT = WORK_DIR / 'coefficient-c5-full'
HIGH_CONFIG = REPOSITORY / 'configs' / 'personal_time_sensitivity.json'
HIGH_POLICY = None  # Optional compatible checked policy input.
GEOMETRY_SUMMARY = WORK_DIR / 'c5-geometry-summary.json'
GEOMETRY_CERTIFICATE = WORK_DIR / 'c5-geometry-certificate.json'
RUN_HIGH_PREPARATION = False
RUN_GEOMETRY_EXPORT = False
high_command = [sys.executable, 'scripts/reproduce_coefficient_diagnostics.py', 'c5',
                '--full', '--output', str(HIGH_OUTPUT), '--configs', str(REPOSITORY / 'configs'),
                '--node', NODE]
if HIGH_POLICY is not None:
    high_command += ['--policy-input', str(HIGH_POLICY)]
print(shlex.join(high_command))
if RUN_HIGH_PREPARATION:
    if HIGH_OUTPUT.exists():
        raise FileExistsError(HIGH_OUTPUT)
    subprocess.run(high_command, cwd=REPOSITORY, check=True)
if RUN_GEOMETRY_EXPORT:
    import json
    from src.design_validation import c5_precision_reproduction, c5_final
    designs, summary, _ = c5_precision_reproduction.load_preparation(HIGH_OUTPUT, REPOSITORY / 'configs')
    certificate = c5_final.certify_supplied_geometry(designs, summary)
    for path, value in ((GEOMETRY_SUMMARY, summary), (GEOMETRY_CERTIFICATE, certificate)):
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('x', encoding='utf-8') as handle:
            json.dump(value, handle, sort_keys=True, allow_nan=False)
```

## Conditional final, under review
This expensive route replays the complete population and checks the high-price producer and independent outputs, fixed inputs and geometry certificates. Keep the high-price output isolated and immutable: the checkpoint lock does not protect that separate directory from other programs. Successful JSON stdout is saved without overwriting an existing file outside the snapshot. Its confidence statement is conditional on the supplied prefix, fixed model, future runtime and independence assumptions; it is not an accepted historical-exact result. Use the same checkpoint location choice as the audit above.

```python
ADOPTED_LOOKUP = STAGED_BUNDLE / 'configs' / 'c5_fixed_initial_policy.json'
LATENT_CONFIG = STAGED_BUNDLE / 'configs' / 'canonical_c5_latent.json'
FINAL_JSON = WORK_DIR / 'conditional-result.json'
RUN_CONDITIONAL_FINAL = False
locations = ([item for path in AUDIT_SHARDS for item in ('--shard-dir', str(path))]
             if AUDIT_SHARDS else ['--output', OUTPUT])
final_command = [*runner, 'final', *population, *locations, '--high-output', str(HIGH_OUTPUT),
                 '--high-config', str(HIGH_CONFIG), '--adopted-lookup', str(ADOPTED_LOOKUP),
                 '--latent-config', str(LATENT_CONFIG), '--geometry-summary', str(GEOMETRY_SUMMARY),
                 '--geometry-certificate', str(GEOMETRY_CERTIFICATE)]
if HIGH_POLICY is not None:
    final_command += ['--policy', str(HIGH_POLICY)]
print(shlex.join(final_command))
if RUN_CONDITIONAL_FINAL:
    FINAL_JSON.parent.mkdir(parents=True, exist_ok=True)
    if FINAL_JSON.exists():
        raise FileExistsError(FINAL_JSON)
    completed = subprocess.run(final_command, cwd=REPOSITORY, check=True, capture_output=True, text=True)
    json.loads(completed.stdout)
    with FINAL_JSON.open('x', encoding='utf-8') as handle:
        handle.write(completed.stdout)
```
