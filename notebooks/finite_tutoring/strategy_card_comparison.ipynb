# Fixed-Design Strategy-Card Comparison

Reproduce six procedural strategies on the same 450 original tutoring roots: equal minutes, similar projected GPA, all to the higher guess, all to the lower guess, ask each once then equalize projected GPA, and a repeated one-step adaptive rule. This is a descriptive comparison, not an adopted human-study target map or proof of global optimality.

Use a separate Python 3.11.5 environment and Node 22.14.0. The card calculation uses NumPy 1.24.3, unlike the earlier personal-time reference. Strict binary64 comparisons can depend on the numerical build; versions alone do not guarantee arbitrary-hardware bitwise identity.

```python
import sys, subprocess, platform
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').is_file())
if platform.python_version() != '3.11.5':
    raise RuntimeError('Select the separate Python 3.11.5 card environment before installation.')
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-e', str(ROOT), 'numpy==1.24.3', 'scipy==1.10.1', 'pandas==2.0.3', 'matplotlib==3.7.5', 'threadpoolctl>=3.1'])
```

## Numerical Scope

The first four cards use rounded visible guesses and ask no questions. The fixed-two card integrates the full 100-by-100 report support and deducts both question fees. Fresh public calculations do not require the historical cached scores. Their source-independent reader retains the original 2e-11 GPA tolerance; fresh values are not labeled historical reused bits.

The adaptive card makes strict binary64 one-step utility comparisons, asks only the chosen recipient, and spends the actual tutoring fee. It has one latent need per physical recipient and one rep0 report stream per root. A separate outward-bounded real-exponential evaluator computes conditional GPA at the chosen terminal allocation. Its numerical intervals are not confidence intervals or estimates of report-stream variation. The five fixed cards have broad payoff-support bounds, not tight numerical-error bars.

Within each condition all 90 roots retain weight 1/90, including unresolved or missing weight. No ranking, within-root Monte Carlo precision or human classification follows from these outputs. The historical literal RNG domain is retained to preserve the same scientific streams.

```python
import json, shlex
RUN = False
RUN_PRODUCTION = False  # Enable only when intentionally generating policy outputs.
FULL = False  # True selects all 450 original roots; rep0 is fixed.
ONLY = 'C3:0:1,C5:0:1'
NODE = 'node'
OUTPUT = ROOT / 'results' / ('strategy_cards_full' if FULL else 'strategy_cards_selected')
SCOPE = ['--full'] if FULL else ['--only', ONLY]
def command(phase, scope):
    return [sys.executable, str(ROOT / 'scripts/reproduce_strategy_cards.py'), phase, '--output', str(OUTPUT), '--node', NODE, *scope]
print(json.dumps(json.loads((ROOT / 'configs/strategy_cards.json').read_text()), indent=2))
```

```python
args = command('prepare', SCOPE)
print(shlex.join(args))
if RUN:
    subprocess.check_call(args, cwd=ROOT)
```

```python
args = command('produce', SCOPE)
print(shlex.join(args))
if RUN and RUN_PRODUCTION:
    subprocess.check_call(args, cwd=ROOT)
```

```python
args = command('check', SCOPE)
print(shlex.join(args))
if RUN:
    subprocess.check_call(args, cwd=ROOT)
```

```python
args = command('aggregate', ['--full'] if FULL else [])
print(shlex.join(args))
if RUN:
    receipt = json.loads(subprocess.check_output(args, cwd=ROOT, text=True))
    summary_path = Path(receipt['summary'])
    print(summary_path.read_text())
    print(summary_path.with_name('condition_cards.csv').read_text())
```

## Outputs and Reuse

The workflow writes original inputs, per-root producer records, independent numerical checks, and the summary JSON/CSV under the ignored output folder. With a prepared full roster, separate calls can select disjoint --only roots in the same output folder. Per-root locks refuse overlaps, and exact completed records are reused. A partial summary keeps its original denominator and never substitutes for the full 30-cell panel.

Leave RUN_PRODUCTION off when checking existing matching outputs. Preparing inputs and checking saved paths do not generate new adaptive trajectories. A newly generated calculation retains its own source/build identity; numerical agreement does not imply identical historical execution.
