## Fresh Computation or Resume
Set RUN=True and FULL=True to reproduce the paper. The default one-root selection verifies the numerical entry but deliberately produces no full paper figures. WORKERS is the number of independent root processes; set it to CPUs available to this calculation. Repeating the same command resumes missing producer/checker records and validates completed ones. Changed sources, runtime, recipe or selection require a separate output directory.

```python
import shlex
RUN = False
FULL = False
WORKERS = 1
OUT = ROOT / 'results' / ('paper_figures' if FULL else 'paper_figures_entry')
command = [sys.executable, str(ROOT / 'scripts/reproduce_paper_figures.py'), 'all', '--output', str(OUT), '--workers', str(WORKERS)]
command += ['--full'] if FULL else ['--only', 'C1:0:1']
print(shlex.join(command))
if RUN:
    subprocess.check_call(command, cwd=ROOT)
```

Use this instead of fresh computation when the six-price producer and independent checker have already finished. No policy is evaluated by reduce. Keep the input separate from the new output. ## Fast Redraw
A complete current-version run contains paper/figure_bundle.json, its checked reductions and action/question tables. This mode verifies that bundle and draws four PNGs and four PDFs, without generating inputs or evaluating policies. It is not an import of arbitrary CSV tables. Changed renderer identities require a new reduction from compatible checked numerical records, not modification of an old seal. Point BUNDLE to either the fresh-run paper directory or the reduce output's paper directory.

```python
RUN_REDRAW = False
BUNDLE = ROOT / 'results' / 'paper_figures' / 'paper'
FIGURES = ROOT / 'results' / 'paper_redraw'
redraw_command = [sys.executable, str(ROOT / 'scripts/reproduce_paper_figures.py'), 'redraw', '--input', str(BUNDLE), '--output', str(FIGURES)]
print(shlex.join(redraw_command))
if RUN_REDRAW:
    subprocess.check_call(redraw_command, cwd=ROOT)
    from IPython.display import Image, display
    for path in sorted(FIGURES.glob('figure_*.png')):
        display(Image(filename=str(path)))
```

## Figures and Interpretation
1. Framework, five environments and observed search/allocation at both 0.001 and 0.01 combined GPA per personal second.
2. Mean absolute share distances to equal split (ES), equal mean surplus (EO), a representative expected-utility-maximizing grid allocation (MU), and all time to lower terminal estimated need (LN). All two-decimal displayed minima receive stars. Distances are not welfare regret or identification of a unique computation.
3. True-state realized utility and the proportion initially asking a question. Personal-time charges are not subtracted from this utility axis.
4. True-need gap and partial lower-need coefficients, oriented to the higher-true-need recipient. Exact true-need ties are excluded. Bars are sample standard deviations across nine fixed schedules, not confidence intervals.
The companion paper/actions/action_conditions.csv and trial_actions.json report the actual terminal routine and paid questions to each displayed slot and logical recipient. These counts come from checked state/events, not allocation resemblance. First display slot is not always logical A. A partial run produces partial tables with explicit denominators.
These are fixed simulated-design descriptions, not population or human inference, continuous price boundaries, or a proof of globally optimal metareasoning. Plotting recipes and definitions are in configs/paper_figures.json; no private result package is needed for fresh reproduction.
