## Model and Identity

Each affordable decision enumerates both 100-report supports. Requesting information reduces the tutoring budget and incurs incremental personal request/read time; terminal computation also has a personal-time cost. The objective is expected combined GPA minus the declared 0.01 GPA/second price. Task GPA itself is not reduced by personal-time accounting. Timing and valuation are explicit reference assumptions, not empirical calibration. Removing the incremental request/read price from a local margin is not a gross-GPA value-of-information estimate.

The original candidate, random-law and native source-seed identifiers remain unchanged. No new policy latent, report or terminal draws are made. Native baseline ancestry generates the initial roots and preserves discarded-fork action-counter gaps; its scientific binary64 and snapshot hashes must match the original identities. The original design's descriptive fields are retained because native fingerprints bind them too.

The independent reader repeats request/stop accounting without calling the producer decision oracle. It shares the accepted finite-task, budget and terminal-macro kernels, so agreement is not an independent arithmetic certificate. Exact discrete decisions and a 1e-10 absolute tolerance for floating values are required.

```python
import json, shlex
RUN = False
FULL = False
RUN_EXAMPLES = False
CONFIG = ROOT / 'configs/personal_time_reference.json'
SELECTION = ROOT / 'configs/personal_time_selection.json'
OUTPUT = ROOT / 'results' / ('personal_time_full' if FULL else 'personal_time_selected')
EXAMPLE_OUTPUT = ROOT / 'results/personal_time_examples'
NODE = 'node'
print(json.dumps(json.loads(CONFIG.read_text()), indent=2))
print('Selected roots:', json.loads(SELECTION.read_text()))
```

## Source-Calling Reproduction

The small selection is C3:0:1 and C5:0:1 and never produces condition-level estimates. Set FULL=True for all 450 original roots; computation remains off until RUN=True. Each mode requires an explicit selection, or --full. The same output folder and selection are used for all stages. Alternatively run the single `all` mode.

Writes are immutable: identical resume preserves existing bytes and completed producer decisions are reused. A changed selection, scientific input, implementation, decision content or output causes refusal rather than replacement. Use a distinct output directory for a distinct selection. Independent checking recomputes every selected decision even on resume.

```python
base = [sys.executable, str(ROOT / 'scripts/reproduce_personal_time.py')]
scope = ['--full'] if FULL else ['--selection', str(SELECTION)]
common = ['--config', str(CONFIG), '--output', str(OUTPUT)] + scope
commands = [base + [mode] + common + (['--node', NODE] if mode == 'prepare' else []) for mode in ('prepare', 'produce', 'check')]
for command in commands:
    print(shlex.join(command))
    if RUN:
        subprocess.check_call(command, cwd=ROOT)
```
