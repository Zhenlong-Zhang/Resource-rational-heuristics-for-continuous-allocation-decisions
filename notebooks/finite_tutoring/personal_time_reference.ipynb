## Model and Identity

Each affordable decision enumerates both 100-report supports. Requesting information reduces the tutoring budget and incurs incremental personal request/read time; terminal computation also has a personal-time cost. The objective is expected combined GPA minus the declared 0.01 GPA/second price. Task GPA itself is not reduced by personal-time accounting. Timing and valuation are explicit reference assumptions, not empirical calibration. Removing the incremental request/read price from a local margin is not a gross-GPA value-of-information estimate.

The original candidate, random-law and native source-seed identifiers remain unchanged. The 0.01 reference section makes no policy latent, report or terminal draws. Native baseline ancestry generates the initial roots and preserves discarded-fork action-counter gaps; its scientific binary64 and snapshot hashes must match the original identities. The original design's descriptive fields are retained because native fingerprints bind them too.

The independent reader repeats request/stop accounting without calling the producer decision oracle. It shares the accepted finite-task, budget and terminal-macro kernels, so agreement is not an independent arithmetic certificate. Exact discrete decisions and a 1e-10 absolute tolerance for floating values are required.

```python
import json, shlex
RUN = False
FULL = False
RUN_EXAMPLES = False
CONFIG = ROOT / 'configs/personal_time_reference.json'
SELECTION = ROOT / 'configs/personal_time_selection.json'
OUTPUT = ROOT / 'results' / ('personal_time_full' if FULL else 'personal_time_selected')
EXAMPLE_OUTPUT = ROOT / 'results/personal_time_examples'
NODE = 'node'
print(json.dumps(json.loads(CONFIG.read_text()), indent=2))
print('Selected roots:', json.loads(SELECTION.read_text()))
```

## Source-Calling Reproduction

The small selection is C3:0:1 and C5:0:1 and never produces condition-level estimates. Set FULL=True for all 450 original roots; computation remains off until RUN=True. Each mode requires an explicit selection, or --full. The same output folder and selection are used for all stages. Alternatively run the single `all` mode.

Writes are immutable: identical resume preserves existing bytes and completed producer decisions are reused. A changed selection, scientific input, implementation, decision content or output causes refusal rather than replacement. Use a distinct output directory for a distinct selection. Independent checking recomputes every selected decision even on resume.

```python
base = [sys.executable, str(ROOT / 'scripts/reproduce_personal_time.py')]
scope = ['--full'] if FULL else ['--selection', str(SELECTION)]
common = ['--config', str(CONFIG), '--output', str(OUTPUT)] + scope
commands = [base + [mode] + common + (['--node', NODE] if mode == 'prepare' else []) for mode in ('prepare', 'produce', 'check')]
for command in commands:
    print(shlex.join(command))
    if RUN:
        subprocess.check_call(command, cwd=ROOT)
```

## Six-Price Comparison

The frozen prices are 0, 0.00001, 0.0001, 0.001, 0.0031622776601683794 and 0.01 combined-GPA units per personal second. They are absolute prices, not multipliers of another reference rate. Request/read timing, the original 20/60-second terminal proxies, the D1000 allocation grid, native roots and stream laws are unchanged. These timings are assumptions, not measured human computation times.

The complete comparison contains 2,700 endpoints: 1,860 expected-only endpoints and 840 realized trajectories. The 0.01 coordinate contributes 450 expected stops, not 450 realized histories. The other five prices use the original paired stream construction and seed 20261019; changing a price does not silently switch to a newer policy or observation model.

Only complete ten-trial participants enter the person-level analysis, and each of the nine schedules receives equal weight in a condition. A proper subset cannot produce the complete finite-condition result. Posterior expected final utility, combined GPA, the priced objective and acquisition behavior are separate fields. The condition averages are finite-stream summaries, not exact ex-ante policy values; within-schedule MCSE is unavailable.

```python
RUN_SENSITIVITY = False
SENSITIVITY_FULL = False
SENSITIVITY_CONFIG = ROOT / 'configs/personal_time_sensitivity.json'
SENSITIVITY_OUTPUT = ROOT / 'results' / ('personal_time_sensitivity_full' if SENSITIVITY_FULL else 'personal_time_sensitivity_selected')
print(json.dumps(json.loads(SENSITIVITY_CONFIG.read_text()), indent=2))
sensitivity_scope = ['--full'] if SENSITIVITY_FULL else ['--selection', str(SELECTION)]
sensitivity_base = [sys.executable, str(ROOT / 'scripts/reproduce_personal_time_sensitivity.py')]
sensitivity_common = ['--config', str(SENSITIVITY_CONFIG), '--output', str(SENSITIVITY_OUTPUT)] + sensitivity_scope
for mode in ('prepare', 'produce', 'check'):
    command = sensitivity_base + [mode] + sensitivity_common + (['--node', NODE] if mode == 'prepare' else [])
    print(shlex.join(command))
    if RUN_SENSITIVITY:
        subprocess.check_call(command, cwd=ROOT)
```

## Sensitivity Results and Limits

Independent checking reconstructs decisions, the recipient-specific observation stream, paid state transitions and terminal metrics before reduction. It shares the finite-task mathematical kernels with the producer; it is not an independent arithmetic certificate. Complete runs write `finite_sensitivity.csv`; `trial_rows.json`, `person_rows.json`, `aggregate.json` and `independent_check.json` preserve denominators and checking scope. Missing or changed inputs, incompatible resume data and incomplete claimed coverage are rejected.

Retain the unfavorable findings: C1 need not choose exactly half at the two lowest prices; at the intermediate 0.0031622776601683794 price C3 asks on 30/90 trials, while C5 asks on 90/90. Scarcity utilities and some heuristic diagnostics are negative; the C5 initial-need slope is undefined. The 0.01 coordinate stops throughout this finite design, not necessarily in other environments. Six tested prices do not locate a continuous threshold or calibrate a human price, and these results do not transfer to a different timing or decision model.
