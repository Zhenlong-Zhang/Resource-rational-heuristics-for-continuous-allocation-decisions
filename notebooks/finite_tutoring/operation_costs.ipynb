# Operation costs and fixed information search

Reproduce literal costed one-more VOC, five precommitted no-question rules, four fixed ask-once/raw policies, and the saved seven-procedure menu with price/time conditions. Use Python 3.11.5, NumPy 1.24.3 and Node 22.14.0, as for the strategy-card workflow. See the [guide](README.md#operation-costs).

The objective is combined GPA minus price times personal seconds. Unknown operation durations remain symbolic; reported rankings use the explicit one-second member, not measured human times. Literal reached-history values are conditional, whereas ask-once integrates all 100 possible reports. No free best-rule or best-direction selector is assumed.


```python
RUN = False
FULL = False
WORKERS = 1
NODE = 'node'
OUTPUT = ROOT / 'results' / ('operation_costs' if FULL else 'operation_costs_small')
CONDITIONS = 'C1,C2,C3,C4,C5' if FULL else 'C5'
FAMILIES = 'noask,literal,ask_once' if FULL else 'noask,ask_once'
DIRECTIONS = 'A,B' if FULL else 'A'
command = [sys.executable, str(ROOT / 'scripts/reproduce_operation_costs.py'), 'all', '--output', str(OUTPUT), '--conditions', CONDITIONS, '--families', FAMILIES, '--directions', DIRECTIONS, '--workers', str(WORKERS), '--node', NODE]
print(' '.join(command))
if RUN:
    subprocess.run(command, cwd=ROOT, check=True)
```
