## Scientific Settings

The frozen recipe enumerates needs/reports 1..100 and share indices 0..1000. Each paid query consumes tutoring resource; personal question time is not priced. Terminal time reduces the decision objective only, never task GPA or tutoring resource. Zero resource has an undefined share and zero terminal duration. Timing profiles are assumptions, not measurements.

Every affordable decision evaluates both full 100-report panels at the reduced budget. Backups use tolerance-selected child objectives, with raw maxima retained separately. Predictive and observed posterior maps remain distinct. Root pathwise bounds take the maximum before expectation and are floating witnesses, not machine-certified bounds. Unresolved contradictions cannot become successful stops.

The fixed namespace `r9-repeated-macro-bridge-v1` and exactly eight potential-stream key fields determine SHA256 -> first 16 bytes -> PCG64 -> first float64 uniform. Scenario is deliberately absent, preserving common uniforms. Each recipient has one fixed latent and its own report counter. Selection never renumbers or reseeds trials.

```python
import json, shlex
RUN = False
CONFIG = ROOT / 'configs/tutoring_cost_aware.json'
SELECTION = ROOT / 'configs/tutoring_cost_aware_selection.json'
OUTPUT = ROOT / 'results/cost_aware_selected'
NODE = 'node'
recipe = json.loads(CONFIG.read_text())
settings = json.loads((CONFIG.parent / recipe['terminal_time']).read_text())
selected = json.loads(SELECTION.read_text())
print(json.dumps({'recipe': recipe, 'timingProfiles': settings['assumed_profiles'], 'selection': selected}, indent=2))
```

## Portable Chain

The selection file contains explicit `conditionId`, `scheduleVariant`, `mainIndex`, `replicateId`, and `scenarioId` coordinates. It illustrates selected trials, not a full-population roster. To choose other supported coordinates, provide another selection JSON without changing their identities. Altering scientific settings is a new model, not reproduction; this interface rejects changed frozen settings.

Each stage requires a fresh output directory. Preparation follows genuine native sessions/checks/practice and baseline ES through each session's last selected main, supporting main indices 1..10. The source is explicitly partial, not a completed session or full panel. Production computes each selected trial once.
```python
base = [sys.executable, str(ROOT / 'scripts/reproduce_cost_aware_tutoring.py')]
common = ['--config', str(CONFIG), '--selection', str(SELECTION)]
commands = [
    base + ['prepare'] + common + ['--output', str(OUTPUT / 'inputs'), '--node', NODE],
    base + ['produce'] + common + ['--inputs', str(OUTPUT / 'inputs'), '--output', str(OUTPUT / 'records')],
    base + ['check'] + common + ['--inputs', str(OUTPUT / 'inputs'), '--records', str(OUTPUT / 'records'), '--output', str(OUTPUT / 'checked'), '--node', NODE],
]
for command in commands:
    print(shlex.join(command))
    if RUN:
        subprocess.check_call(command, cwd=ROOT)
```

## Read the Selected Evidence

`inputs/inputs.json` records the complete scientific context and exact native root log bits. `records/<trial_name>.ndjson` contains trial, decision, latent, paid and history rows; decision rows embed all report blocks. `records/reference.json`, written last, retains roots, the symmetric C1 reference rate, final histories and SHA256 of each raw public row excluding its newline. These are public transport identities, not historical manifests.

`checked/verification.json` reports only checked selections, numerical errors and completion. Both exact recorded-operand and independently reconstructed full100 backups/policies are checked, including discrete action/tie agreement. The checker derives its price from its own reference calculation. Independent numerical agreement uses field-specific tolerances; exact map/draw/backup checks are not relaxed. The implementations share NumPy and PCG64, so agreement is not an independent arithmetic certificate or proof of historical execution. Expected utility, expected combined GPA, net objective and realized values are distinct.

```python
if RUN:
    reference = json.loads((OUTPUT / 'records/reference.json').read_text())
    verification = json.loads((OUTPUT / 'checked/verification.json').read_text())
    print(json.dumps(reference['reference'], indent=2))
    for trial in reference['trials']:
        history = trial['history']
        fields = ('status', 'reason', 'paidCount', 'finalBudget', 'shareIndex', 'terminalSeconds', 'expectedUtility', 'expectedCombinedGPA', 'expectedObjective', 'realized')
        print(json.dumps({'selection': trial['selection'], **{k: history[k] for k in fields}}, indent=2))
    print(json.dumps(verification, indent=2))
```

## Complete Study Reproduction

The complete route uses five conditions, nine native schedule variants, 64 model-person replicates per variant and ten ordered main trials, for each of three timing profiles and six price multipliers. The small mode below contains one ten-main person and cannot establish full-study completion.

The public plan regenerates native source roots, partitions only file ownership, computes and independently checks each selected history, then reduces complete persons. It preserves original initial predictors, response-independent projection weights, equal-person eligibility and variant-stratified Monte Carlo standard errors. H1/H2 use the deterministic initial asking map, not mean question counts. Assigned-person and standardized-variant routes are reported separately; the latter has no Monte Carlo SE.

Use the reference environment above. Initial-slope plans are regenerated in an isolated single-thread MKL 2021.4 AVX subprocess; its effective mode and numerical binaries are recorded. This does not change the parent policy runtime or claim universal bitwise portability. No historical output download is required. The producer and independent numerical checker have distinct implementations but share NumPy/PCG64. Full reproduction can be expensive; choose worker count for available memory and CPUs. Shard size does not change seeds or denominators.


```python
RUN_STUDY = False
STUDY_MODE = 'smoke'  # Set 'full' for the complete original study.
PEOPLE_PER_SHARD = 8
WORKERS = 1
STUDY_OUTPUT = ROOT / 'results/cost_aware_study'
SHARD_IDS = None  # None runs all; or a list of stable ordinals, e.g. [0, 1].
study_runner = [sys.executable, str(ROOT / 'scripts/reproduce_cost_aware_study.py')]
prepare_command = study_runner + ['prepare', '--mode', STUDY_MODE, '--people-per-shard', str(PEOPLE_PER_SHARD), '--output', str(STUDY_OUTPUT / 'prepared'), '--node', NODE]
run_command = study_runner + ['run', '--prepared', str(STUDY_OUTPUT / 'prepared'), '--output', str(STUDY_OUTPUT / 'computed'), '--workers', str(WORKERS), '--node', NODE]
run_command += ['--all'] if SHARD_IDS is None else ['--ids', ','.join(map(str, SHARD_IDS))]
summary_command = study_runner + ['aggregate', '--prepared', str(STUDY_OUTPUT / 'prepared'), '--output', str(STUDY_OUTPUT / 'computed'), '--summary', str(STUDY_OUTPUT / 'summary.json'), '--node', NODE]
for command in (prepare_command, run_command, summary_command):
    print(shlex.join(command))
    if RUN_STUDY:
        subprocess.check_call(command, cwd=ROOT)
```

## Results And Interpretation

`prepared/plan.json` fixes the population and shard coordinates. Each `computed/shards/<ordinal>/` contains source inputs, raw decisions/observations/histories, independent replay and a file-bound completion marker. Re-running **run** reuses verified shards and completed trial files, and resumes missing checking or completion stages only when their inputs and records match. Conflicting artifacts are rejected. A genuinely incomplete trial file must be preserved separately before retrying that trial; it is never overwritten. Do not rerun **prepare** over an existing destination. Choose a new summary filename for each readback.

`summary.json` reports exact missing ordinals. It creates full study aggregates only when all original shards are complete: 90 performance rows, 36 initial-asking contrasts and 2,052 other original endpoint/route rows. An incomplete or smoke study never gets full-study status. Returned identities are public reproduction identities, not historical execution receipts.

Expected utility remains the performance measure. Expected combined GPA, the time-priced net objective, realized values and information-acquisition measures are separate. Original H6/H7 expected-GPA resemblance is retained at candidate scope; the later prospective surplus-primary mapping does not relabel these historical rows. High-cost zero or reversed contrasts remain results, not errors to tune away. Personal request/read time is omitted in this model; timing values are assumptions. These finite-grid predictions neither determine human behavior nor establish a continuous rate boundary.
