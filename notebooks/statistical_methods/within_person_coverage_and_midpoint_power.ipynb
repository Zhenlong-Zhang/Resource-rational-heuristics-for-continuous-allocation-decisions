## Fixed Settings

All computation switches default off. Selected IDs retain their original full-population identities and denominators; they are not a newly sized study.
```python
import json
import shlex

SETTINGS = json.loads((ROOT / 'configs/within_statistics.json').read_text())
RUN = False
FULL = False
RUN_MAPPING = False
RUN_SAVED_SCALARS = False
RUN_COMMON_T = False
RSCRIPT = 'Rscript'
NODE = 'node'
POLICY = ROOT / 'results/personal_time_sensitivity_full'
ALLOCATIONS = ROOT / 'results/allocation_welfare_full'
MAPPING = ROOT / 'results/proxy_h10_mapping'
SUFFIX = 'full' if FULL else 'selected'
COVERAGE = ROOT / ('results/within_coverage_' + SUFFIX)
MIDPOINT = ROOT / ('results/midpoint_power_' + SUFFIX)
STAT = ROOT / 'scripts/reproduce_within_statistics.py'

for study in ('coverage', 'midpoint'):
    plan = within_design.plan(study)
    assert plan['expectedTotal'] == SETTINGS[study]['full_count']
    print(study, plan['expectedTotal'], SETTINGS[study]['selected_original_ordinals'])

def invoke(command, enabled):
    print(shlex.join([str(arg) for arg in command]))
    if enabled:
        subprocess.run([str(arg) for arg in command], cwd=ROOT, check=True)
```

## Common-T Paired Comparison

This is a separate retrospective method comparison on the same S1/S5 coverage inputs, not the midpoint power scenario. First complete the selected or full coverage produce/check stages above, or select compatible saved public coverage outputs. This stage reads the saved within-person estimates and their independent checks; it never generates observations, refits REML/CR2, recomputes the within estimator or changes its linear intervals.

The candidate uses an exact, predictor-only sign certificate for the covariance of D and S. Opposite observed signs with certified nonnegative covariance use a common-df Student tail; negative or valid uncertified signs retain the original within-IUT2 p-value. Same/zero observed signs return 1. Strict Holm4 and the original fixed family remain unchanged. An independent pairwise-Gram/beta-tail implementation checks the numerical result.

The accepted historical comparison had S1 family R/F 0/10,000 for both methods and S5 R/D 0/2,129 versus 5/2,129. Paired tables (00,01,10,11) were (10000,0,0,0) and (2124,5,0,0). The five gains are three A13 and two B13 rejections. This small retrospective gain is not a between-method significance result, prospective calibration or human power. S1 D and S5 F are inapplicable, not zero. The adverse S5 within coverage and sub-80% midpoint result are unchanged.

Keep the upstream source checkout used to generate coverage. If its original location changed, set COVERAGE_SOURCE_ROOT to the preserved checkout; every recorded source file must still match exactly. This explicitly locates old source bytes without relabeling historical execution. Missing or modified prerequisites are refused, not regenerated.


```python
from tutoring_statistics import common_t_reproduction
COMMON_SETTINGS = json.loads((ROOT / 'configs/common_t_comparison.json').read_text())
COMMON = ROOT / ('results/common_t_comparison_' + SUFFIX)
COMMON_CLI = ROOT / 'scripts/reproduce_common_t.py'
COVERAGE_SOURCE_ROOT = None
assert common_t_reproduction.plan()['methodID'] == COMMON_SETTINGS['methodID']
assert common_t_reproduction.plan()['expectedTotal'] == COMMON_SETTINGS['full_count']
archive = ['--coverage-source-root', COVERAGE_SOURCE_ROOT] if COVERAGE_SOURCE_ROOT is not None else []
invoke([sys.executable, COMMON_CLI, 'prepare', '--coverage', COVERAGE,
        '--output', COMMON, *archive], RUN_COMMON_T)
common_ids = ','.join(map(str, COMMON_SETTINGS['selected_original_ordinals']))
common_selector = ['--all'] if FULL else ['--ids', common_ids]
for stage in ('produce', 'check'):
    invoke([sys.executable, COMMON_CLI, stage, '--output', COMMON,
            *common_selector, '--execute'], RUN_COMMON_T)
invoke([sys.executable, COMMON_CLI, 'reduce', '--output', COMMON,
        *(['--full'] if FULL else []), '--execute'], RUN_COMMON_T)
if RUN_COMMON_T:
    filename = 'aggregate_full.json' if FULL else f'aggregate_partial_{len(COMMON_SETTINGS["selected_original_ordinals"])}.json'
    print(json.dumps(json.loads((COMMON / filename).read_text())['report'], indent=2))
```
